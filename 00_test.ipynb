```python
# Tic-Tac-Toe

board = [" " for _ in range(9)]


def display_board():
    print()
    print(f" {board[0]} | {board[1]} | {board[2]} ")
    print("---+---+---")
    print(f" {board[3]} | {board[4]} | {board[5]} ")
    print("---+---+---")
    print(f" {board[6]} | {board[7]} | {board[8]} ")
    print()


def check_winner(player):
    winning_positions = [
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),
        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),
        (0, 4, 8),
        (2, 4, 6)
    ]

    for position in winning_positions:
        first, second, third = position

        if (
            board[first] == player
            and board[second] == player
            and board[third] == player
        ):
            return True

    return False


def board_is_full():
    return " " not in board


current_player = "X"

print("Welcome to Tic-Tac-Toe!")
print("Choose a position from 1 to 9.")
print("The positions are arranged like this:")
print()
print(" 1 | 2 | 3 ")
print("---+---+---")
print(" 4 | 5 | 6 ")
print("---+---+---")
print(" 7 | 8 | 9 ")

while True:
    display_board()

    try:
        choice = int(input(f"Player {current_player}, choose a position: "))
    except ValueError:
        print("Please enter a number from 1 to 9.")
        continue

    if choice < 1 or choice > 9:
        print("Invalid position. Choose a number from 1 to 9.")
        continue

    position = choice - 1

    if board[position] != " ":
        print("That position is already occupied.")
        continue

    board[position] = current_player

    if check_winner(current_player):
        display_board()
        print(f"Player {current_player} wins!")
        break

    if board_is_full():
        display_board()
        print("The game is a draw!")
        break

    if current_player == "X":
        current_player = "O"
    else:
        current_player = "X"
```

```python
purity = float(input("Enter the purity of the product (as a percentage): "))
print(f"the purity of the product is {purity}%")

pH= float(input("Enter the pH of the product: "))
print(f"The pH of the product is {pH}")

if purity >=98.0 or 6.5 <= pH <= 7.5:
    classification = "Grade A"
elif purity >=95.0 or 6.0<= pH <= 8.0:
    classification = "Grade B"
elif purity >=90.0 or 5.0 <= pH <= 9.0:
    classification = "Grade C"
else:
    classification = "Rejected"
print(f"The product is classified as: {classification}")
```
