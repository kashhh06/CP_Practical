how to use math.exp() to calculate the exact value of C intial at 10 min and print absolute error

```python
#question no. 2
#part 2
import math

C0 = 2.00
k = 0.08
delta_t = 1
t = 10

# Euler approximation
C_approx = C0
for i in range(t):
    C_approx = C_approx - k * C_approx * delta_t

C_exact = C0 * math.exp(-k * t)

# Absolute error
abs_error = abs(C_exact - C_approx)

print(f"Exact C at {t} min = {C_exact:.10f}")
print(f"Approximate C at {t} min = {C_approx:.10f}")
print(f"Absolute error = {abs_error:.10f}")
```

give code for sample no. fibonacci tag and sampling time using fibonaaci iterative and recursive


```python
# Fibonacci sequence: F0 = 0, F1 = 1

def fib_iter(n):
    if n <= 0:
        return []
    if n == 1:
        return [0]

    fib_list = [0, 1]
    while len(fib_list) < n:
        fib_list.append(fib_list[-1] + fib_list[-2])
    return fib_list


def fib_reck(k):
    if k == 0:
        return 0
    if k == 1:
        return 1
    return fib_reck(k - 1) + fib_reck(k - 2)

n = 10
iter_values = fib_iter(n)
recursive_values = [fib_reck(i) for i in range(n)]

if iter_values == recursive_values:
    print("Both functions agree.")
else:
    print("Mismatch detected!")
    print("Iterative:", iter_values)
    print("Recursive:", recursive_values)

# print table: sample no | fibonacci tag | sampling time (min)
print("\nSample No. | Fibonacci Tag | Sampling Time (min)")
print("-" * 45)

for sample_no, fib_tag in enumerate(iter_values, start=1):
    sampling_time = fib_tag * 2
    print(f"{sample_no:<10} | {fib_tag:<13} | {sampling_time}")
```
